```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.primitives import StatevectorSampler
```

```python
plus = Statevector.from_label("+").data
minus = Statevector.from_label("-").data

estado = Statevector(
    (np.sqrt(3) / 2) * plus
    + (1 / 2) * minus
)
```

```python
circuito = QuantumCircuit(1)
circuito.h(0)

total = 100000
mas = 0
menos = 0
for i in range(total):
    nuevo_estado = estado.evolve(circuito)
    resultado, _ = nuevo_estado.measure([0])
    if resultado == '0':
        mas += 1
    else:
        menos += 1

print(f'Fracción de resultados asociados a +: {mas/total}')
print(f'Fracción de resultados asociados a -: {menos/total}')
```
