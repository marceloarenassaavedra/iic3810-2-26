```python
from qiskit import QuantumCircuit
```

```python
def circuito_entrada(entrada):
    n = len(entrada)
    circuito = QuantumCircuit(2*n-1)
    for i in range(n):
        if entrada[i] == 0:
            circuito.x(n-1-i)
    circuito.ccx(0,1,n)
    for i in range(2,n):
        circuito.ccx(i,i+n-2,i+n-1)
    for i in range(n-2,1,-1):
        circuito.ccx(i,i+n-2,i+n-1)
    circuito.ccx(0,1,n)
    for i in range(n):
        if entrada[i] == 0:
            circuito.x(n-i-1)
    return(circuito)
```

```python
def circuito_tabla(tabla, dim):
    circuito = QuantumCircuit(2*dim-1)
    for i in range(len(tabla)):
        circuito = circuito.compose(circuito_entrada(tabla[i]))
        if i < len(tabla)-1:
            circuito.barrier()
    return circuito
```

```python
tabla = [ [0,1,0,1], [0,1,0,0], [1, 1, 1, 1] ]
dim = 4
circuito = circuito_tabla(tabla, dim)
circuito.draw("mpl")
```
