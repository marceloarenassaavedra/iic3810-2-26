```python
import numpy as np
from qiskit import QuantumCircuit

def qft(circuito, qubits):
    n = len(qubits)

    for s in range(n - 1, -1, -1):
        circuito.h(qubits[s])

        for p in range(s - 1, -1, -1):
            angulo = np.pi / (2 ** (s - p))
            circuito.cp(
                angulo,
                qubits[s],
                qubits[p]
            )
        circuito.barrier()
        
    for s in range(n // 2):
        circuito.swap(qubits[s], qubits[n - 1 - s])
```

```python
t = 3
Q = 2 ** t
total = 1 + t

circuito = QuantumCircuit(total, t)

circuito.x(0)

circuito.barrier()

for s in range(1, total):
    circuito.h(s)

circuito.barrier()

for s in range(0, t):
    angulo = 2 * np.pi * (3 / 8) * (2 ** s)
    circuito.cp(angulo, 1 + s, 0)

circuito.barrier()

circuito_qft = QuantumCircuit(t)
qft(circuito_qft, list(range(t)))
circuito.compose(
    circuito_qft.inverse(),
    qubits=[1, 2, 3],
    inplace=True
)

circuito.barrier()

circuito.measure([1, 2, 3], [0, 1, 2])

circuito.draw("mpl")
```

```python
from qiskit.primitives import StatevectorSampler

simulador = StatevectorSampler()
resultado = simulador.run(
    [circuito],
    shots=1000
).result()

conteos = resultado[0].data.c.get_counts()
print(conteos)
```
