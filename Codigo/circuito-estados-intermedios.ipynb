```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
circuito = QuantumCircuit(2)

estado = Statevector.from_instruction(circuito)
print(f'Estado inicial             : {estado.probabilities_dict()}')

circuito.h(0)
circuito.id(1)

estado = Statevector.from_instruction(circuito)
print(f'Estado después de I x H    : {estado.probabilities_dict()}')

circuito.cx(0, 1)

estado = Statevector.from_instruction(circuito)
print(f'Estado después de CNOT     : {estado.probabilities_dict()}')

resultado, estado = estado.measure([0])

print(f'Resultado de la medición   : {resultado}')
print(f'Estado después de medición : {estado.probabilities_dict()}')
```
