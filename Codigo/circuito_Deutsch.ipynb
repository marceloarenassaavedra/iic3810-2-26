```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
```

```python
def oraculo(v0, v1):
    circuito = QuantumCircuit(2)

    if v0 == 0:
        if v1 == 1:
            circuito.cx(0, 1)
    else:
        if v1 == 0:
            circuito.x(1)
            circuito.cx(0,1)
        else:
            circuito.x(1)
        
    return circuito
```

```python
def circuito_deutsch(oraculo):
    circuito = QuantumCircuit(2, 1)

    circuito.x(1)
    circuito.h(0)
    circuito.h(1)
    circuito.barrier()
    circuito.compose(oraculo, inplace=True)
    circuito.barrier()
    circuito.h(0)
    circuito.measure(0, 0)

    return circuito
```

```python
for v0, v1 in [(0,0), (0,1), (1,0), (1,1)]:
    U_f = oraculo(v0, v1)
    D_f = circuito_deutsch(U_f)

    simulador = StatevectorSampler()
    resultado = simulador.run(
        [D_f],
        shots=1000
    ).result()

    display(D_f.draw("mpl"))
    conteos = resultado[0].data.c.get_counts()
    print(conteos)
```
